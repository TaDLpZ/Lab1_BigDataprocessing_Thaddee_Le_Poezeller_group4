```python
# Initialize a word counter by creating a tuple with word and cound of 1
words = rdd.flatMap(lambda lines: lines.split(' ')) \
                    .map(lambda word: (word, 1))

for w in words.collect():
    print(w)
```

```python
# a. count the occurence of each word
word_counts = rdd.flatMap(lambda lines: lines.split(' ')) \
                 .map(lambda word: (word, 1)) \
                 .reduceByKey(lambda a, b: a + b)

for w in word_counts.take(20):
    print(w)
```

```python
# b. a common first step in text analysis, change all capital letters to lower case
words_lower = rdd.flatMap(lambda lines: lines.lower().split(' ')) \
                    .map(lambda word: (word, 1))

words_lower.take(20)
```

```python
stop_words = {
    'a', 'an', 'the', 'and', 'or', 'but', 'of', 'to', 'in', 'on', 'at', 'by',
    'for', 'with', 'from', 'as', 'is', 'was', 'were', 'be', 'been', 'are',
    'it', 'its', 'he', 'she', 'his', 'her', 'him', 'they', 'them', 'their',
    'we', 'you', 'i', 'me', 'my', 'that', 'this', 'these', 'those', 'which',
    'who', 'what', 'not', 'no', 'so', 'if', 'then', 'than', 'had', 'has',
    'have', 'do', 'did', 'would', 'could', 'there', 'all', 'said', ''
}

words_clean = rdd.flatMap(lambda line: line.lower().split(' ')) \
                 .filter(lambda word: word not in stop_words) \
                 .map(lambda word: (word, 1))

words_clean.take(20)
```

```python
# d. sort in alphabetical order
words_sort = rdd.flatMap(lambda line: line.lower().split(' ')) \
                 .filter(lambda word: word not in stop_words) \
                 .map(lambda word: (word, 1))\
                 .sortByKey(lambda word: (word, 1))

words_sort.take(20)

# in our case punctuated word appear first i will do it again after the punctuation part
```

```python
# e. sort descending by word frequency

words_frequency = rdd.flatMap(lambda line: line.lower().split(' ')) \
                 .filter(lambda word: word not in stop_words) \
                 .map(lambda word: (word, 1))\
                 .reduceByKey(lambda a, b: a + b)\
                 .sortBy(lambda x: x[1], ascending=False)

words_frequency.take(20)
```

```python
# f. remove punctuations and blank spaces
import re

words_nopunct = rdd.flatMap(lambda line: re.findall(r"[a-z']+", line.lower())) \
                 .filter(lambda word: word not in stop_words) \
                 .map(lambda word: (word, 1))

words_nopunct.take(20)

#i will test the sortby with the no ponctuation
```

## 3. Function timing.

- write a simple python timer function for seeing how quickly your rdd runs as written. change the order of the steps in order to make the rdd run as optimally as possible


```python
# Code here
import time
def time_it(func):
    start = time.perf_counter()
    result = func()
    elapsed = time.perf_counter() - start
    print(elapsed)
    return result

time_it(lambda: agesRDD.take(20))
```
